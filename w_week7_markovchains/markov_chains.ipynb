# Markov Chains

```python
import numpy as np
import random as rm
```

```python
# The statespace
states = ["Sleep","Taco Bell","Gym"]

# Possible sequences of events
transitionName = [["SS","SG","ST"],["GS","GG","GT"],["TS","TG","TT"]]

# Probabilities matrix (transition matrix)
transitionMatrix = [[0.2,0.6,0.2],[0.1,0.6,0.3],[0.2,0.7,0.1]]
```

$\sum_{j=1}^{m} p_{ij} = 1$, so we need to ensure that each row adds up to 1.

Creat a function that simulates the daily activities.

 Starting from **Sleep**, each day they randomly transition to one of three states: Sleep, Gym, or Taco Bell — based on fixed probabilities in the`transitionMatrix`.

| From \ To | Sleep | Gym | Taco Bell |
|-----------|-------|-----|-----------|
| **Sleep** | 0.2 | 0.6 | 0.2 |
| **Gym** | 0.2 | 0.5 | 0.3 |
| **Taco Bell** | 0.2 | 0.7 | 0.1 |

`activity_forecast(days)` returns a list of activities for each day (e.g. `["Sleep", "Gym", "Taco Bell"]`), and tracks the cumulative probability of that exact path.

The simulation then runs **10,000 times** over 2 days to empirically estimate the probability of going from **Sleep → Gym** in 2 steps.

```python
def activity_forecast(days):
    # Choose the starting state
    activityToday = "Sleep"
    activityList = [activityToday]
    i = 0
    prob = 1
    while i != days:
        if activityToday == "Sleep":
            change = np.random.choice(transitionName[0],replace=True,p=transitionMatrix[0])
            if change == "SS":
                prob = prob * 0.2
                activityList.append("Sleep")
                pass
            elif change == "SG":
                prob = prob * 0.6
                activityToday = "Gym"
                activityList.append("Gym")
            else:
                prob = prob * 0.2
                activityToday = "Taco Bell"
                activityList.append("Taco Bell")
        elif activityToday == "Gym":
            change = np.random.choice(transitionName[1],replace=True,p=transitionMatrix[1])
            if change == "GG":
                prob = prob * 0.5
                activityList.append("Gym")
                pass
            elif change == "GS":
                prob = prob * 0.2
                activityToday = "Sleep"
                activityList.append("Sleep")
            else:
                prob = prob * 0.3
                activityToday = "Taco Bell"
                activityList.append("Taco Bell")
        elif activityToday == "Taco Bell":
            change = np.random.choice(transitionName[2],replace=True,p=transitionMatrix[2])
            if change == "TT":
                prob = prob * 0.1
                activityList.append("Taco Bell")
                pass
            elif change == "TS":
                prob = prob * 0.2
                activityToday = "Sleep"
                activityList.append("Sleep")
            else:
                prob = prob * 0.7
                activityToday = "Gym"
                activityList.append("Gym")
        i += 1    
    return activityList

# To save every activityList
list_activity = []
count = 0

# `Range` starts from the first count up until but excluding the last count
for iterations in range(1,10000):
        list_activity.append(activity_forecast(2))

# Check out all the `activityList` we collected    
#print(list_activity)

# Iterate through the list to get a count of all activities ending in state 'Gym'
for smaller_list in list_activity:
    if(smaller_list[2] == "Gym"):
        count += 1

# Calculate the probability of starting 'Sleep' and ending at 'Gym'
prob = (count/10000)
print("The probability of starting at state 'Sleep' and ending at state 'Gym'= " + str(prob))
```
