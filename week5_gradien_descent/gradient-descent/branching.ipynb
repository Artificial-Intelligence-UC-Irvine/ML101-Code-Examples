## Introduction to Gradient Descent

Gradient Descent is an optimization algorithm that:

- Finds the minimum of a function by iteratively moving in the direction of steepest descent
- Is the foundation of most machine learning training algorithms
- Uses derivatives (gradients) to determine the direction to move
- Adjusts parameters step by step to minimize a cost/loss function

We'll implement:
- Linear regression using gradient descent
- Mean Squared Error (MSE) cost function
- Batch gradient descent algorithm
- Visualization of the optimization process

## Gradient Descent Implementation

**Time to build our optimizer!** 

The Gradient Descent algorithm works by:
1. **Initialize**: Start with random or zero weights and bias
2. **Predict**: Make predictions using current parameters
3. **Calculate Cost**: Measure how wrong our predictions are (using MSE)
4. **Calculate Gradients**: Find which direction to move to reduce cost
5. **Update Parameters**: Move a small step in that direction
6. **Repeat**: Keep doing steps 2-5 until cost stops decreasing

**Key concept**: The gradient tells us the slope - we move DOWN the slope to find the minimum!
Let's implement this step by step!


```python
class LinearRegressionGD:
    def __init__(self, learning_rate=0.01, max_iterations=1000, tolerance=1e-6):
        """
        Initialize Linear Regression with Gradient Descent.
        
        Parameters:
        learning_rate: Step size for gradient descent
        max_iterations: Maximum number of iterations
        tolerance: Convergence threshold
        """
        self.learning_rate = learning_rate
        self.max_iterations = max_iterations
        self.tolerance = tolerance
        self.weights = None
        self.bias = None
        self.cost_history = []
        print("🏗️  LinearRegressionGD class initialized!")
        print(f"Learning rate: {learning_rate}")
        print("This will store our learned weights, bias, and training history.")

    def fit(self, X, y):
        """Train the model using gradient descent."""
        print("📚 Starting gradient descent training...")
        # Initialize parameters
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0
        self.cost_history = []
        
        print(f"Initial weights: {self.weights}, Initial bias: {self.bias}")
        print("Starting gradient descent loop...\n")
        
        # Gradient descent loop
        for i in range(self.max_iterations):
            # Predictions
            y_pred = X.dot(self.weights) + self.bias
            
            # Calculate cost (MSE)
            cost = np.mean((y_pred - y) ** 2)
            self.cost_history.append(cost)
            
            # Calculate gradients
            dw = (2 / n_samples) * X.T.dot(y_pred - y)
            db = (2 / n_samples) * np.sum(y_pred - y)
            
            # Update parameters
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db
            
            # Print progress every 50 iterations
            if (i + 1) % 50 == 0:
                print(f"Iteration {i+1}: Cost = {cost:.4f}, Weight = {self.weights[0]:.4f}, Bias = {self.bias:.4f}")
            
            # Check for convergence
            if i > 0 and abs(self.cost_history[-2] - self.cost_history[-1]) < self.tolerance:
                print(f"\n✅ Converged after {i+1} iterations!")
                break
        
        print(f"\n🎯 Training complete! Final cost: {self.cost_history[-1]:.4f}")
        return self

    def predict(self, X):
        """Make predictions."""
        return X.dot(self.weights) + self.bias

    def score(self, X, y):
        """Calculate R-squared score."""
        y_pred = self.predict(X)
        ss_res = np.sum((y - y_pred) ** 2)
        ss_tot = np.sum((y - np.mean(y)) ** 2)
        return 1 - (ss_res / ss_tot)

print("🎉 LinearRegressionGD class defined!")
print("This class implements gradient descent for linear regression.")
```

```python
# Train the model
print("🚀 Training our model...")
model = LinearRegressionGD(learning_rate=0.003, max_iterations=400)
model.fit(X, y)

print(f"\n📊 Final Model Parameters:")
print(f"Learned weight: {model.weights[0]:.4f}")
print(f"Learned bias: {model.bias:.4f}")
print(f"Final cost: {model.cost_history[-1]:.4f}")
print(f"R-squared score: {model.score(X, y):.4f}")
print("\n💭 The model learned the best line to fit our data!")
```

## Visualizing the Results

Let's see how well our model fits the data!


```python
# Plot the data and fitted line
plt.figure(figsize=(10, 6))
plt.scatter(X, y, alpha=0.6, color='blue', s=50, label='Data points')
X_line = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, 'r-', linewidth=2, label='Fitted line')
plt.xlabel('Feature (X)', fontsize=12)
plt.ylabel('Target (y)', fontsize=12)
plt.title('Linear Regression Fit', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("🎨 The red line shows our learned model!")
print("Notice how it passes through the center of the data points.")
```

```python
# Plot cost history
plt.figure(figsize=(10, 6))
plt.plot(range(len(model.cost_history)), model.cost_history, 'b-', linewidth=2)
plt.xlabel('Iteration', fontsize=12)
plt.ylabel('Cost (MSE)', fontsize=12)
plt.title('Cost Function Over Iterations', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("📉 Watch how the cost decreases over time!")
print("This shows gradient descent is successfully minimizing our error.")
```

## Testing Different Learning Rates

Let's see how different learning rates affect the optimization process!


```python
# Test different learning rates
print("🧪 Testing different learning rates...")
learning_rates = [0.001, 0.003, 0.005, 0.008]
colors = ['blue', 'green', 'orange', 'red']

plt.figure(figsize=(12, 8))
for lr, color in zip(learning_rates, colors):
    print(f"  Training with learning rate {lr}...")
    model_test = LinearRegressionGD(learning_rate=lr, max_iterations=400)
    model_test.fit(X, y)
    plt.plot(range(len(model_test.cost_history)), model_test.cost_history, 
             color=color, linewidth=2, label=f'LR={lr}')

plt.xlabel('Iteration', fontsize=12)
plt.ylabel('Cost (MSE)', fontsize=12)
plt.title('Cost Function with Different Learning Rates', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\n💡 Notice how:")
print("- Smaller learning rates converge more slowly but smoothly")
print("- Larger learning rates converge faster but still show gradual improvement")
print("- The cost decreases steadily over many iterations, showing the optimization process!")
```
