## Indexing
Similar to indexing in a Python list object, you can index a numpy array. 

### Basic Indexing
This is how you can access any data point or feature of your `X`. Each index corresponds to an entry.

```python
# first row of our data
meeting_X[0]
```

```python
# first two rows
meeting_X[0:2]
```

```python
# another way to say first row
meeting_X[0, :]
```

```python
# first column
meeting_X[:,0]
```

```python
# first target label of first data point
meeting_y[0]
```

```python
# set a value
meeting_y[0] = 3
```

```python
meeting_y[0]
```

### Indexing via Slicing
Python "slicing" allows us to access entire rows or columns. We can also access contiguous rows & columns.

```python
# first 5 rows of meeting_X
meeting_X[:5, :]
```

```python
# the 10th to *14th* rows of meeting_X
meeting_X[10:15, :]
```

```python
# the 1st to 3rd rows and 1st to 3rd columns (features)
meeting_X[1:4, 1:4]
```

```python
# the last row of meeting_X
meeting_X[-1, :]
```

```python
# the last 3 rows of meeting_X
meeting_X[-3:, :]
```

```python
# first three columns
meeting_X[:, :-1]
```

```python
# get rows 1, 5, 9 from meeting_X
fetch_rows = [1, 5, 9]
meeting_X[fetch_rows, :]
```

```python
# get 1st and last column from meeting_X
fetch_cols = [0, -1]
meeting_X[:, fetch_cols]
```

### Logical Indexing
You can index an array depending on specifications of the array's content. This allows for a boolean to be returned and for us to inspect parts of the data that follow some sort of binary "rule."

```python
bday_months = meeting_X[:,2] # the 2nd column is the birthday month feature
bday_months # all birthday months
```

```python
# how many board members are born in the latter half of the year?
bday_months > 6
```

```python
# what are the bday months of those born in the latter half of the year?
bday_months[bday_months > 6]
```

```python
# who are the board members born in the latter half of the year?
meeting_X[bday_months > 6]
```

```python
upperclassmen = meeting_y > 2 # how many board members are upperclassmen?
upperclassmen
```

```python
# who is an upperclassmen?
meeting_y[upperclassmen]
```

```python
meeting_X[upperclassmen]
```

```python
# how many members are upperclassmen?
len(meeting_X[upperclassmen])
```

### Feature Statistics (Optional)
Numpy has a number of functions you can use to find out the statistics of your data, primarily for you input (X) data.

```python
bday_months = meeting_X[:,2] # the 2nd column is the birthday month feature
bday_months # all birthday months
```

```python
# what is the average bday month?
np.mean(meeting_X[:,2])
```

```python
# what are the mean, variance, standard deviation, min, and max of bday months?
print(f'Mean: {np.mean(meeting_X[:,2], axis=0)}') # axis = 0 means going up and down the column; axis = 1 means going left and right
print(f'Variance: {np.var(meeting_X[:,2])}')
print(f'Standard Deviation: {np.std(meeting_X[:,2])}')
print(f'Minimum: {min(meeting_X[:,2])}')
print(f'Maximum: {max(meeting_X[:,2])}')
```

```python
# what are the mean, variance, standard deviation, min, and max of bday months for board members who like Windows?
print(f'Mean: {np.mean(meeting_X[meeting_X[:,3] == 'Windows'][:,2], axis=0)}') # axis = 0 means going up and down the column; axis = 1 means going left and right
print(f'Variance: {np.var(meeting_X[meeting_X[:,3] == 'Windows'][:,2])}')
print(f'Standard Deviation: {np.std(meeting_X[meeting_X[:,3] == 'Windows'][:,2])}')
print(f'Minimum: {min(meeting_X[meeting_X[:,3] == 'Windows'][:,2])}')
print(f'Maximum: {max(meeting_X[meeting_X[:,3] == 'Windows'][:,2])}')
```

```python
# what are the mean, variance, standard deviation, min, and max of bday months for board members who like MacOS?
print(f'Mean: {np.mean(meeting_X[meeting_X[:,3] == 'MacOS'][:,2], axis=0)}') # axis = 0 means going up and down the column; axis = 1 means going left and right
print(f'Variance: {np.var(meeting_X[meeting_X[:,3] == 'MacOS'][:,2])}')
print(f'Standard Deviation: {np.std(meeting_X[meeting_X[:,3] == 'MacOS'][:,2])}')
print(f'Minimum: {min(meeting_X[meeting_X[:,3] == 'MacOS'][:,2])}')
print(f'Maximum: {max(meeting_X[meeting_X[:,3] == 'MacOS'][:,2])}')
```

```python
# the mean of an entire row
print(f'Mean: {np.mean(meeting_X[0], axis=0)}')
print(f'Standard Deviation: {np.std(meeting_X[0], axis=0)}')
```
