### Linspace and Logspace

```python
## x-space create a sequence that starts at base ** start (base to the power of start) and ends with base ** stop
lin = np.linspace(1.0, 4.0, 4) # base is 1
plt.scatter(range(4), lin)
plt.title("Linspace Function")
plt.show()
print(lin)
log = np.logspace(0.0, 2.0, 4) # base is 10
plt.scatter(range(4), log)
plt.title("Logspace Function")
plt.show()
print(log)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12.0, 3.0))

# hist for BirthdayMonth
ax[0].hist(df['BirthdayMonth'], bins=12, color='skyblue', edgecolor='black')
ax[0].set_xlabel('Birthday Month')
ax[0].set_ylabel('Count')
ax[0].set_title('Distribution of Birthday Months')

# hist for Favorite OS (encoded)
ax[1].hist(df['FaveOSModified'], bins=np.arange(-0.5, 3.5, 1), color='lightgreen', edgecolor='black', rwidth=0.8)
ax[1].set_xticks([0,1,2])
ax[1].set_xticklabels(['Linux','Windows','MacOS'])
ax[1].set_xlabel('Favorite OS')
ax[1].set_ylabel('Count')
ax[1].set_title('Distribution of Favorite OS')

plt.tight_layout()
plt.show()
```
